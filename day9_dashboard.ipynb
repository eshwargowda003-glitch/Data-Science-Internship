```python
import pandas as pd
import matplotlib.pyplot as plt

# Display plots inside notebook
%matplotlib inline
```

```python
df = pd.read_csv("dataset/students.csv")

print("Student Dataset")
df
```

```python
print("Total Students :", df["ID"].count())
print("Average Age :", round(df["Age"].mean(),2))
print("Average CGPA :", round(df["CGPA"].mean(),2))
print("Minimum CGPA :", df["CGPA"].min())
print("Maximum CGPA :", df["CGPA"].max())

print("\nStudents by Department")
print(df["Department"].value_counts())
```

```python
plt.figure(figsize=(8,5))

plt.bar(df["Name"], df["CGPA"])

plt.title("Student CGPA")
plt.xlabel("Students")
plt.ylabel("CGPA")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(df["Name"], df["Age"], marker="o")

plt.title("Student Age")
plt.xlabel("Students")
plt.ylabel("Age")

plt.show()
```

```python
department = df["Department"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    department,
    labels=department.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Students by Department")

plt.show()
```

```python
print("Mini Dashboard Completed Successfully!")

print("\nBusiness Insights")
print("--------------------------")
print("• Total Students :", len(df))
print("• Highest CGPA :", df["CGPA"].max())
print("• Lowest CGPA :", df["CGPA"].min())
print("• Average CGPA :", round(df["CGPA"].mean(),2))
print("• Most Students belong to:")
print(df["Department"].value_counts().idxmax())
```
